```python
import sys
sys.path.append("../src")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from xgboost import XGBRegressor
```

```python

from ml.preprocessing import (
    clean_train_pipeline,
    get_feature_columns,
    add_rolling_features,
    add_rolling_slope,
)

from ml.training import(
    evaluate_model,
    build_error_analysis,
    split_by_engine,
    prepare_train_val_data
)
```

```python
train = clean_train_pipeline(
    "../data/train_FD001.txt"
)
train.shape
```

```python
train_data, val_data = split_by_engine(train)

print("Train:", train_data.shape)
print("Validation:", val_data.shape)
```

```python
train_engines = set(train_data["unit_number"])
val_engines = set(val_data["unit_number"])

print("Overlap:", train_engines.intersection(val_engines))
```

```python
X_train, y_train, X_val, y_val = prepare_train_val_data(train_data, val_data, feature_cols=get_feature_columns(train_data))
print("X_train:", X_train.shape)
print("y_train:", y_train.shape)
print("X_val:", X_val.shape)
print("y_val:", y_val.shape)
```

```python
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_val_scaled = scaler.transform(X_val)
```

```python
print(X_train_scaled.shape)
print(X_val_scaled.shape)
```

```python
y_pred = model.predict(X_val)

print(y_pred[:10])
```

```python
baseline_results = evaluate_model(y_val, y_pred, "Random Forest")
```

## Rolling Features

```python
rolling_sensors = [
    "sensor_11",
    "sensor_4",
    "sensor_12",
    "sensor_7"
]

train_rolling = add_rolling_features(
    train,
    sensors=rolling_sensors,
    windows=[5, 10]
)
```

```python
train_rolling.tail(10)
```

```python
train_rolling.head(10)
```

```python
train_rolling.isna().sum().sort_values(ascending=False).head(20)
```

## Add Rolling Feature to train data

```python
train_rolling_data, val_rolling_data = split_by_engine(
    train_rolling
)

print("Train:", train_rolling_data.shape)
print("Validation:", val_rolling_data.shape)

print(
    "Engine overlap:",
    set(train_rolling_data["unit_number"])
    & set(val_rolling_data["unit_number"])
)
```

```python
feature_cols_rolling = get_feature_columns(train_rolling)

X_train_rolling, y_train_rolling, X_val_rolling, y_val_rolling = (
    prepare_train_val_data(
        train_rolling_data,
        val_rolling_data,
        feature_cols_rolling
    )
)

print("Number of features:", len(feature_cols_rolling))
print("X_train:", X_train_rolling.shape)
print("X_val:", X_val_rolling.shape)
print("y_train:", y_train_rolling.shape)
print("y_val:", y_val_rolling.shape)
```

```python
rolling_xgb = XGBRegressor(
    n_estimators=200,
    learning_rate=0.03,
    max_depth=4,
    min_child_weight=3,
    subsample=1.0,
    colsample_bytree=0.8,
    random_state=42,
    n_jobs=-1
)

rolling_xgb.fit(
    X_train_rolling,
    y_train_rolling
)

y_pred_rolling = rolling_xgb.predict(
    X_val_rolling
)
```

```python
rolling_results = evaluate_model(
    y_val_rolling,
    y_pred_rolling,
    model_name="XGBoost + Rolling Features"
)
```

```python
rolling_error_analysis = build_error_analysis(
    val_rolling_data,
    y_pred_rolling
)

rolling_error_analysis.sort_values(
    "absolute_error",
    ascending=False
).head(10)
```

```python
rolling_error_analysis.describe()
```

```python
slope_xgb = XGBRegressor(
    n_estimators=200,
    learning_rate=0.03,
    max_depth=4,
    min_child_weight=3,
    subsample=1.0,
    colsample_bytree=0.8,
    random_state=42,
    n_jobs=-1
)

slope_xgb.fit(
    X_train_slope,
    y_train_slope
)

y_pred_slope = slope_xgb.predict(
    X_val_slope
)
```

```python
slope_results = evaluate_model(
    y_val_slope,
    y_pred_slope,
    model_name="XGBoost + Rolling + Slope"
)
```
