```python
import sys
sys.path.append("../src")

from ml.preprocessing import (
    clean_train_pipeline,
    get_feature_columns,
    add_rolling_features,
    cap_rul
)

from ml.training import(
    split_by_engine,
    prepare_train_val_data,
    evaluate_model,
    build_error_analysis,
)

from config import ROLLING_SENSORS,ROLLING_WINDOWS

from xgboost import XGBRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from itertools import product
```

```python
train = clean_train_pipeline(
    "../data/train_FD001.txt"
)
train.shape
```

```python
train_data, val_data = split_by_engine(train)

print("Train:", train_data.shape)
print("Validation:", val_data.shape)
```

```python
X_train, y_train, X_val, y_val = prepare_train_val_data(train_data, val_data, feature_cols=get_feature_columns(train_data))

print("X_train:", X_train.shape)
print("y_train:", y_train.shape)
print("X_val:", X_val.shape)
print("y_val:", y_val.shape)
```

```python
param_combinations = [
    {"n_estimators": 200, "learning_rate": 0.05, "max_depth": 4},
    {"n_estimators": 200, "learning_rate": 0.05, "max_depth": 6},
    {"n_estimators": 200, "learning_rate": 0.05, "max_depth": 8},
    {"n_estimators": 300, "learning_rate": 0.05, "max_depth": 4},
    {"n_estimators": 300, "learning_rate": 0.05, "max_depth": 6},
    {"n_estimators": 300, "learning_rate": 0.05, "max_depth": 8},
]
```

```python
results = []

for params in param_combinations:

    model = XGBRegressor(
        **params,
        random_state=42,
        n_jobs=-1
    )

    model.fit(X_train, y_train)

    pred = model.predict(X_val)

    mae = mean_absolute_error(y_val, pred)
    rmse = np.sqrt(mean_squared_error(y_val, pred))
    r2 = r2_score(y_val, pred)

    results.append({
        "n_estimators": params["n_estimators"],
        "learning_rate": params["learning_rate"],
        "max_depth": params["max_depth"],
        "MAE": mae,
        "RMSE": rmse,
        "R2": r2
    })
```

```python
tuning_results = pd.DataFrame(results)

tuning_results.sort_values(
    "MAE",
    ascending=True
)
```

## Learning Rate

```python
learning_rates = [0.03, 0.05, 0.10]

results_lr = []

for lr in learning_rates:

    model = XGBRegressor(
        n_estimators=200,
        learning_rate=lr,
        max_depth=4,
        random_state=42,
        n_jobs=-1
    )

    model.fit(X_train, y_train)

    pred = model.predict(X_val)

    mae = mean_absolute_error(y_val, pred)
    rmse = np.sqrt(mean_squared_error(y_val, pred))
    r2 = r2_score(y_val, pred)

    results_lr.append({
        "learning_rate": lr,
        "MAE": mae,
        "RMSE": rmse,
        "R2": r2
    })

learning_rate_results = pd.DataFrame(results_lr)

learning_rate_results.sort_values(
    "MAE",
    ascending=True
)
```

## Final tuning check with rolling features with Grid search

```python
train_rolling = add_rolling_features(
    train,
    sensors=ROLLING_SENSORS,
    windows=ROLLING_WINDOWS
)

feature_cols_rolling = get_feature_columns(train_rolling)

print("Shape:", train_rolling.shape)
print("Number of features:", len(feature_cols_rolling))
```

```python
train_rolling_data, val_rolling_data = split_by_engine(
    train_rolling
)

X_train_rolling, y_train_rolling, X_val_rolling, y_val_rolling = (
    prepare_train_val_data(
        train_rolling_data,
        val_rolling_data,
        feature_cols_rolling
    )
)

print("X_train:", X_train_rolling.shape)
print("X_val:", X_val_rolling.shape)
print("Engine overlap:", set(
    train_rolling_data["unit_number"]
).intersection(
    val_rolling_data["unit_number"]
))
```

```python
param_grid = {
    "n_estimators": [150, 200, 250],
    "learning_rate": [0.02, 0.03, 0.04],
    "max_depth": [3, 4, 5],
    "min_child_weight": [1, 3],
    "subsample": [0.8, 1.0],
    "colsample_bytree": [0.8, 1.0],
}

param_combinations = list(product(
    param_grid["n_estimators"],
    param_grid["learning_rate"],
    param_grid["max_depth"],
    param_grid["min_child_weight"],
    param_grid["subsample"],
    param_grid["colsample_bytree"],

))

print("Number of combinations:", len(param_combinations))
```

```python
results = []

for (
    n_estimators,
    learning_rate,
    max_depth,
    min_child_weight,
    subsample,
    colsample_bytree
) in param_combinations:

    model = XGBRegressor(
        n_estimators=n_estimators,
        learning_rate=learning_rate,
        max_depth=max_depth,
        min_child_weight=min_child_weight,
        subsample=subsample,
        colsample_bytree=colsample_bytree,
        random_state=42,
        n_jobs=-1
    )

    model.fit(
        X_train_rolling,
        y_train_rolling
    )

    y_pred = model.predict(
        X_val_rolling
    )

    mae = mean_absolute_error(
        y_val_rolling,
        y_pred
    )

    rmse = np.sqrt(
        mean_squared_error(
            y_val_rolling,
            y_pred
        )
    )

    r2 = r2_score(
        y_val_rolling,
        y_pred
    )

    results.append({
        "n_estimators": n_estimators,
        "learning_rate": learning_rate,
        "max_depth": max_depth,
        "min_child_weight": min_child_weight,
        "subsample": subsample,
        "colsample_bytree": colsample_bytree,
        "MAE": mae,
        "RMSE": rmse,
        "R2": r2
    })

grid_results = pd.DataFrame(results)

print("Completed:", len(grid_results))
```
