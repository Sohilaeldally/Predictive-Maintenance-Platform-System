```python
import pandas as pd

test = pd.read_csv(
    "../data/test_FD001.txt",
    sep=r"\s+",
    header=None
)

test.shape
```

```python
columns = [
    "unit_number",
    "time_cycles",
    "op_setting_1",
    "op_setting_2",
    "op_setting_3",
]

columns += [f"sensor_{i}" for i in range(1, 22)]

test.columns = columns
```

```python
test.groupby("unit_number")["time_cycles"].max().reset_index().head(10)
```

```python
rul_test = pd.read_csv(
    "../data/RUL_FD001.txt",
    header=None
)

rul_test
```

```python
last_cycles = (
    test.groupby("unit_number")["time_cycles"]
    .max()
    .reset_index()
)

last_cycles.columns = ["unit_number", "last_cycle"]

last_cycles["true_RUL"] = rul_test[0]

last_cycles["failure_cycle"] = (
    last_cycles["last_cycle"] + last_cycles["true_RUL"]
)

last_cycles.head(10)
```
